### 3. Setup Ngrok Auth Token
Sign up at [ngrok.com](https://dashboard.ngrok.com/signup) to get a free authtoken. Paste it below when prompted.

```python
import getpass
from pyngrok import ngrok, conf

print("Enter your ngrok authtoken:")
ngrok_token = getpass.getpass()
conf.get_default().auth_token = ngrok_token
```

### 4. Start Streamlit & Expose App
Run this cell to start Streamlit and get your public URL.

```python
import os

# Run Streamlit in the background
get_ipython().system_raw('streamlit run app.py &>/content/logs.txt &')

# Expose port 8501 via ngrok
public_url = ngrok.connect(8501)
print("\n🌍 Your Streamlit App is running at:", public_url)
print("\nClick the link above to open the chatbot. (Note: downloading the 7B model weights will take a few minutes when you first ask a question!)")
```
